```python
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
import kagglehub
from model import GazeEstimationNet
from dataset import MPIIGazeDataset
from train import denormalize_batch, collect_predictions, plot_test_results

device       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DROPOUT_RATE = 0.5
SAVE_DIR     = '/content/checkpoints'

stats = torch.load(f'{SAVE_DIR}/ensemble_stats.pth', map_location=device, weights_only=False)
dataset_path = kagglehub.dataset_download("dhruv413/mpiigaze")

test_dataset = MPIIGazeDataset(
    dataset_path, ['p13', 'p14'], split='test',
    samples_per_participant=3000,
    gaze_ranges=stats['gaze_ranges'],
    img_mean=stats['img_mean'],
    img_std=stats['img_std'],
)
test_loader = torch.utils.data.DataLoader(
    test_dataset, batch_size=128, shuffle=False, num_workers=2
)

all_preds = []
for i in range(1, 6):
    m = GazeEstimationNet(DROPOUT_RATE).to(device)
    m.load_state_dict(torch.load(f'{SAVE_DIR}/model_fold_{i}.pth', map_location=device))
    preds, targets = collect_predictions(m, test_loader, test_dataset, device)
    all_preds.append(preds)

ensemble_preds = torch.stack(all_preds).mean(dim=0)
pred_d = denormalize_batch(ensemble_preds, test_dataset)
tgt_d  = denormalize_batch(targets, test_dataset)

per_sample_err = np.array([
    np.degrees(np.arccos(np.clip(np.dot(
        np.array([-np.sin(p[0])*np.cos(p[1]), -np.sin(p[1]), -np.cos(p[0])*np.cos(p[1])]),
        np.array([-np.sin(t[0])*np.cos(t[1]), -np.sin(t[1]), -np.cos(t[0])*np.cos(t[1])]),
    ), -1, 1)))
    for p, t in zip(pred_d.numpy(), tgt_d.numpy())
])

plot_test_results(pred_d, tgt_d, per_sample_err)
plt.show()


def draw_gaze_arrow(img_bgr, yaw, pitch, color, length=120, thickness=2):
    h, w = img_bgr.shape[:2]
    cx, cy = w // 2, h // 2
    cv2.arrowedLine(img_bgr, (cx, cy),
                    (int(cx - length*np.sin(yaw)), int(cy - length*np.sin(pitch))),
                    color, thickness, tipLength=0.3, line_type=cv2.LINE_AA)

best_model = GazeEstimationNet(DROPOUT_RATE).to(device)
best_model.load_state_dict(torch.load(f'{SAVE_DIR}/model_fold_1.pth', map_location=device))
best_model.eval()

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()
fig.suptitle("Gaze Demo  |  Red = Ground Truth   Green = Predicted",
             fontsize=13, fontweight="bold")

for plot_i, idx in enumerate(np.random.choice(len(test_dataset), 8, replace=False)):
    img_tensor, gt_norm, pose_tensor, _ = test_dataset[idx]
    with torch.no_grad():
        pred_norm = best_model(img_tensor.unsqueeze(0).to(device),
                               pose_tensor.unsqueeze(0).to(device))
    pred_gaze = test_dataset.denormalize_gaze(pred_norm.cpu().numpy()[0])
    gt_gaze   = test_dataset.denormalize_gaze(gt_norm.numpy())

    img_np  = np.clip(img_tensor.squeeze().numpy() * test_dataset.img_std + test_dataset.img_mean, 0, 1)
    img_bgr = cv2.cvtColor((img_np * 255).astype(np.uint8), cv2.COLOR_GRAY2BGR)
    img_bgr = cv2.resize(img_bgr, (img_bgr.shape[1]*5, img_bgr.shape[0]*5),
                         interpolation=cv2.INTER_CUBIC)

    draw_gaze_arrow(img_bgr, gt_gaze[0],   gt_gaze[1],   color=(0, 0, 255))
    draw_gaze_arrow(img_bgr, pred_gaze[0], pred_gaze[1], color=(0, 255, 0))

    axes[plot_i].imshow(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))
    axes[plot_i].axis("off")
    axes[plot_i].set_title(f"Error: {per_sample_err[idx]:.1f}°", fontsize=11)

plt.tight_layout()
plt.savefig("gaze_demo.png", dpi=150, bbox_inches="tight")
plt.show()
```
